```python
#Product-wise sales summary
# Calculate total units sold, total revenue, and average revenue for each product.
```

```python
df.groupby("Product").agg(
    Total_Units=("Units_Sold", "sum"),
    Total_Revenue=("Revenue", "sum"),
    Average_Revenue=("Revenue", "mean")
)
```

```python
# Region-wise performance
# Compare total revenue, average order revenue, and number of orders in each region.
```

```python
df.groupby("Region").agg(
    Total_Revenue=("Revenue", "sum"),
    Average_Revenue=("Revenue", "mean"),
    Order_Count=("Order_ID", "count")
).sort_values("Total_Revenue", ascending=False)
```

```python
# Category-wise sales analysis
# Find total units sold, total revenue, and the highest order revenue for each category.
```

```python
df.groupby("Category").agg(
    Total_Units=("Units_Sold", "sum"),
    Total_Revenue=("Revenue", "sum"),
    Highest_Order=("Revenue", "max")
)
```

```python
# Monthly revenue analysis
# Group the data by month and calculate total revenue, average revenue, and total units sold.
```

```python
df.groupby(df["Order_Date"].dt.to_period("M")).agg(
    Total_Revenue=("Revenue", "sum"),
    Average_Revenue=("Revenue", "mean"),
    Total_Units=("Units_Sold", "sum")
)
```

```python
# Product-wise, region-wise analysis
# Find total revenue and total units sold for each product in each region.
```

```python
df.groupby(["Product", "Region"]).agg(
    Total_Revenue=("Revenue", "sum"),
    Total_Units=("Units_Sold", "sum"),
    Order_Count=("Order_ID", "count")
).sort_values("Total_Revenue", ascending=False)
```

```python
# Calculate monthly revenue -- line chart
```

```python
monthly_revenue = df.groupby("Month")["Revenue"].sum().sort_index()

plt.figure(figsize=(10, 5))
plt.plot(monthly_revenue.index, monthly_revenue.values, marker="o")

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue (₹)")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Bar chart — Revenue by product
# Compares the total revenue generated by each product.
```

```python
product_revenue = df.groupby("Product")["Revenue"].sum().sort_values(ascending=False)

product_revenue.plot(kind="bar", figsize=(10, 5), title="Revenue by Product")

plt.xlabel("Product")
plt.ylabel("Revenue (₹)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Histogram — Distribution of order revenue
# Shows how frequently order revenues fall within different ranges.
```

```python
df["Revenue"].plot(
    kind="hist",
    bins=6,
    figsize=(10, 5),
    title="Distribution of Order Revenue"
)

plt.xlabel("Revenue (₹)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
# Scatter plot — Units sold vs. unit price
# Examines the relationship between the number of units sold and the price per unit.
```

```python
df.plot(
    kind="scatter",
    x="Units_Sold",
    y="Unit_Price",
    marker="o",
    figsize=(8, 5),
    title="Units Sold vs Unit Price"
)

plt.xlabel("Units Sold")
plt.ylabel("Unit Price (₹)")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
